```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("Student_data.csv")
print(df.head())
```

#drop unmamed column

```python
df=df.drop("Unnamed: 0" , axis = 1)
```

```python
print(df.head())
```

```python
df.head()
```

#no other transformation is required now lets start buildig the charts and start analysis.

#Gender Distribution.

```python
plt.figure(figsize=(5,5))

ax = sns.countplot(
    data=df,
    x="Gender",
    hue="Gender",
    palette={"female": "coral", "male": "skyblue"},
    legend=False
)

for container in ax.containers:
    ax.bar_label(container, padding=3, fontsize=9)
plt.title("Gender Distribution")
plt.show()
```

#from the above chart we have analysed that the number of females in the data is more than the number of males.

```python
gb = df.groupby("ParentEduc").agg({"MathScore":'mean' , "ReadingScore":'mean' , "WritingScore":'mean'})
print(gb)
```

#here is this data we can see that one values are string and other set of values are numeric so here we will represent this as heat map.

```python
plt.figure(figsize=(5,5))
sns.heatmap(gb , annot = True)
plt.title("Relationship between Parent's Education and Student's Score")
plt.show()
```

#From the above chart we have concluded that the education of parents have a good impact on their child's scores.

```python
gb1 = df.groupby("ParentMaritalStatus").agg({"MathScore":'mean' , "ReadingScore":'mean' , "WritingScore":'mean'})
print(gb1)
```

```python
sns.heatmap(gb1 , annot = True)
plt.title("Relationship between Parent's Marital Status and Scores")
plt.show()
```


#From the above chart we have concluded that the marital status of parents have no impact or negligible impact on their child's scores.

#Distribition of data according to the Ethnic groups.

```python
print(df["EthnicGroup"].unique())
```

 Distribution of Ethnic Groups.

```python
groupA=df.loc[(df["EthnicGroup"]=="group A")].count()
groupB=df.loc[(df["EthnicGroup"]=="group B")].count()
groupC=df.loc[(df["EthnicGroup"]=="group C")].count()
groupD=df.loc[(df["EthnicGroup"]=="group D")].count()
groupE=df.loc[(df["EthnicGroup"]=="group E")].count()

l=["group A","group B","group C","group D","group E"]
mlist=[groupA["EthnicGroup"],groupB["EthnicGroup"],groupC["EthnicGroup"],groupD["EthnicGroup"],groupE["EthnicGroup"]]

plt.pie(mlist , labels = l , autopct = "%1.2f%%")
plt.title("Distribution of Ethnic Groups")
plt.show()
```

#here we have used pandas .loc function to calculate the count of every group in ethnic column and then prepared the pie chart and given the labels and used the autopct function to calculate the percentage.
